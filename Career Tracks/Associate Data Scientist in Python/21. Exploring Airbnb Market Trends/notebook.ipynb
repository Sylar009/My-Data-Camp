Welcome to New York City, one of the most-visited cities in the world. There are many Airbnb listings in New York City to meet the high demand for temporary lodging for travelers, which can be anywhere between a few nights to many months. In this project, we will take a closer look at the New York Airbnb market by combining data from multiple file types like `.csv`, `.tsv`, and `.xlsx`.

Recall that **CSV**, **TSV**, and **Excel** files are three common formats for storing data. 
Three files containing data on 2019 Airbnb listings are available to you:

**data/airbnb_price.csv**
This is a CSV file containing data on Airbnb listing prices and locations.
- **`listing_id`**: unique identifier of listing
- **`price`**: nightly listing price in USD
- **`nbhood_full`**: name of borough and neighborhood where listing is located

**data/airbnb_room_type.xlsx**
This is an Excel file containing data on Airbnb listing descriptions and room types.
- **`listing_id`**: unique identifier of listing
- **`description`**: listing description
- **`room_type`**: Airbnb has three types of rooms: shared rooms, private rooms, and entire homes/apartments

**data/airbnb_last_review.tsv**
This is a TSV file containing data on Airbnb host names and review dates.
- **`listing_id`**: unique identifier of listing
- **`host_name`**: name of listing host
- **`last_review`**: date when the listing was last reviewed

```python
# Import necessary packages
import pandas as pd
import numpy as np

# Begin coding here ...
# Use as many cells as you like
```

```python
# Read the CSV file into a DataFrame
df_price = pd.read_csv('data/airbnb_price.csv')

# Read the Excel file into a DataFrame
df_room_type = pd.read_excel('data/airbnb_room_type.xlsx')

# Read the TSV file into a DataFrame
df_last_review = pd.read_csv('data/airbnb_last_review.tsv', sep='\t')

# Display the first few rows of each DataFrame to take a look at the data
print(df_price.head())
print(df_room_type.head())
print(df_last_review.head())
```

```python
# Merge the DataFrames on 'listing_id'
df_merged = df_price.merge(df_room_type, on='listing_id').merge(df_last_review, on='listing_id')

# Display the first few rows of the merged DataFrame to take a look at the combined data
df_merged.head()
```

```python
# Convert the 'last_review' column to datetime format
df_merged['last_review'] = pd.to_datetime(df_merged['last_review'])

# Find the earliest and most recent review dates
earliest_review = df_merged['last_review'].min()
most_recent_review = df_merged['last_review'].max()

# Display the earliest and most recent review dates
print(earliest_review)
print(most_recent_review)
```

```python
# Convert to lowercase and strip whitespace
df_merged['room_type'] = df_merged['room_type'].str.lower().str.strip()

# Count how many rows have 'private room'
private_room_count = (df_merged['room_type'] == 'private room').sum()

# Display the result
print(f"Number of private room listings: {private_room_count}")
```

```python
# Remove ' dollars', strip whitespace, and convert to float
df_merged['price'] = df_merged['price'].str.replace(' dollars', '').str.replace(',', '').astype(float)

# Finding the mean
average_price = df_merged['price'].mean()
print(f"Average price of listings: ${average_price:.2f}")
```

```python
# Round average price to 2 decimal places
average_price_rounded = round(average_price, 2)

# Create the DataFrame with correct column names and order
review_dates = pd.DataFrame({
    'first_reviewed': [earliest_review],
    'last_reviewed': [most_recent_review],
    'nb_private_rooms': [private_room_count],
    'avg_price': [average_price_rounded]
})

# Display the DataFrame
print(review_dates)
```
